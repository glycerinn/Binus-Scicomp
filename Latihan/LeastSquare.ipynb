```python
import numpy as np
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 9]
y = [2, 5, 1, 7, 9, 3, 6, 10]

x = np.array(x)
y = np.array(y)

X = np.vstack([x, np.ones(len(x))]).T

alpha = np.dot(np.dot(np.linalg.inv(np.dot(X.T, X)), X.T), y)

plt.plot(x, y, 'ob')
plt.plot(x, alpha[0]*x + alpha[1], '-r')
plt.title('least square r')
plt.show 
```
