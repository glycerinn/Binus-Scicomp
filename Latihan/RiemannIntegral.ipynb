```python
import numpy as np

def f(x):
    return x**2 - 5

a = 2
b = 5

n = 5

sqSize = (b-a)/(n-1)

x = np.linspace(a, b, n)
y = f(x)

leftR = sqSize * np.sum(y[:n-1])
rightR = sqSize * np.sum(y[1:])

midx = (x[:n-1] + x[1:])/2
midy = f(midx)
midr = sqSize * np.sum(midy)

parside = y[0] + 2 * np.sum(y[1:n-1]) + y[n-1]
trapr = sqSize * parside/2

print(f'left rie is {leftR}')
print(f'right rie is {rightR}')
print(f'mid rie is {midr}')
print(f'trap rie is {trapr}')
```
