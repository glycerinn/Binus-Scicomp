```python
import numpy as np

def diagdom(x):
    x = np.array(x)

    diag = np.diag(np.abs(x))
    offdiag = np.sum(np.abs(x), axis = 1) - diag

    if np.all (diag > offdiag):
        return True
    else:
        return False
    
def gauss(x, y, tol = 0.022, iter= 15):
    print(f'A: {x}, Y:{y}')
    if not diagdom(x):
        print('not dominant')
        return
    
    x = np.array(x)
    y = np.array(y)

    diag = np.diag((x))
    x = -x

    np.fill_diagonal(x, 0)
    old = np.zeros(np.shape(x[0]))

    for i in range(iter):
        new = np.array(old)

        for idx, row in enumerate(x):
            new[idx] = ((y[idx] + np.dot(new, row))/diag[idx])
        print(f'iter {i+1}: {new}')

        dx = np.sqrt(np.dot(new - old, new - old))

        if(dx < tol):
            print('converged')
            return
        old = new
    
    print('not converged')
    return
```

```python
Xs = [
    [
      [4, 2, -1],
      [1, -5, 2],
      [2, -1, -4]
    ],
    [
      [3, 4, 5],
      [-3, 7, -4],
      [1, -4, -2]
    ],
    [
      [9, -2, 3, 2],
      [2, 8, -2, 3],
      [-3, 2, 11, -4],
      [-2, 3, 2, 10]
    ]
]
Ys = [
    [41, -10, 1],
    [34, -32, 62],
    [55, -14, 12, -21]
]

for i, (x, y) in enumerate(zip(Xs, Ys)):
    print(f'equation number {i+1}')
    gauss(x, y)
    print('====================')
```
