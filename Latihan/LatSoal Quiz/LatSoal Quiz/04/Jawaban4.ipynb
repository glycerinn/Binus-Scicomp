```python
import numpy as np

def f(x):
    return x**6 + 2*x**2 - 3

def g(x):
    return 6*x**5 + 4*x

def newtonRaphson(x0, tol = 0.01, iter = 15):
    for i in range(iter):
        x1 = x0 - (f(x0)/g(x0))

        print(f'iteration of {i+1}. new root = {x1}')
        if(f(x1) < tol):
            print('the root of the equation is', x1)
            return
        x0 = x1

newtonRaphson(4)
```
