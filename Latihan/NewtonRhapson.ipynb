```python
import numpy as np

def f(x):
    return x**6 + 2*x**2 - 3

def g(x):
    return 6*x**5 + 4*x

def newtonR(x0 , tol, iter = 15):
    for i in range(iter):
        x1 = x0 - (f(x0)/g(x0))

        print(f'iterasi {i+1}, the root is now {x1}')
        if(f(x1)<tol):
            print('the root of equation is', x1)
            return

        x0=x1

newtonR(4, 0.1)
```
